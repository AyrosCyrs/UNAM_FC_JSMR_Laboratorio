## 3. Valores lógicos

Aquí entramos ya en terreno del Tema I. En R los valores de verdad se escriben
`TRUE` y `FALSE`, siempre en mayúsculas. Los operadores disponibles son los que
esperarías:

| Operador | Significado |
|---|---|
| `&` | conjunción |
| `\|` | disyunción |
| `!` | negación |
| `xor()` | disyunción exclusiva |
| `==` | equivalencia |

Lo interesante es que estos operadores también son vectorizados, lo cual será la
clave de toda la sección siguiente.

```r
p <- TRUE
q <- FALSE

p & q
p | q
!p
xor(p, q)
p == q
```

### Implicación

R no incluye un operador para la implicación lógica, pero eso no es un problema:
la equivalencia $p \rightarrow q \equiv \neg p \lor q$ nos permite definirlo en
una línea. De paso, esta es nuestra primera función.

```r
implica <- function(p, q) {
  (!p) | q
}

implica(TRUE, TRUE)
implica(TRUE, FALSE)
implica(FALSE, TRUE)
implica(FALSE, FALSE)
```

Fíjate en los paréntesis alrededor de `!p`. Aquí no alteran el resultado, pero
conviene adquirir el hábito, porque la precedencia de operadores en R guarda un
par de sorpresas. Las dos que más vale la pena tener presentes son que `&` se
evalúa antes que `|`, y que `!` se evalúa *después* que `==`.

```r
p <- FALSE
q <- FALSE
r <- TRUE

# Como & tiene prioridad sobre |, estas dos expresiones no son la misma.
p & q | r      # R lo lee como (p & q) | r
p & (q | r)    # y esto es otra cosa
```

```r
x <- 5
y <- 3

# Aquí la comparación se resuelve primero y la negación se aplica al final.
!x == y     # R lo lee como !(x == y)
(!x) == y   # en cambio aquí se niega x antes de comparar
```

Mientras trabajes sólo con valores lógicos la segunda trampa rara vez cambia el
resultado, pero en cuanto haya números de por medio empieza a morder. La regla
práctica es simple: si en una expresión conviven negaciones con comparaciones o
con varios conectivos distintos, escribe los paréntesis aunque parezcan
redundantes.

## 4. Tablas de verdad

Construir una tabla de verdad en R es sorprendentemente breve. La función
`expand.grid()` genera todas las combinaciones posibles de los valores que le
pases, es decir, exactamente el producto cartesiano que necesitamos.

```r
tabla <- expand.grid(p = c(TRUE, FALSE),
                     q = c(TRUE, FALSE))
tabla
```

Lo que acabas de obtener es un `data.frame`: una tabla con columnas nombradas,
que es la estructura de datos más usada en R. Puedes acceder a cada columna con
el signo `$`.

Como los operadores lógicos están vectorizados, agregar la columna de resultados
para cualquier fórmula se reduce a escribir la fórmula tal cual.

```r
tabla$conjuncion <- tabla$p & tabla$q
tabla$disyuncion <- tabla$p | tabla$q
tabla$implicacion <- implica(tabla$p, tabla$q)

tabla
```

Con tres variables el procedimiento es idéntico; sólo cambia el número de
renglones, que pasa de cuatro a ocho.

```r
tabla3 <- expand.grid(p = c(TRUE, FALSE),
                      q = c(TRUE, FALSE),
                      r = c(TRUE, FALSE))

tabla3$formula <- (tabla3$p & tabla3$q) | (!tabla3$r)
tabla3
```

## 5. Tautologías, contradicciones y equivalencias

Una vez que tenemos la columna de resultados, decidir si una fórmula es
tautología o contradicción es cuestión de revisar esa columna completa. Las
funciones `all()` y `any()` hacen justamente eso: la primera indica si **todos**
los valores son verdaderos, la segunda si hay **al menos uno**.

```r
es_tautologia <- function(columna) {
  all(columna)
}

es_contradiccion <- function(columna) {
  !any(columna)
}

# Ley del tercero excluido: p o no p.
tercero_excluido <- tabla$p | (!tabla$p)
es_tautologia(tercero_excluido)

# Principio de no contradicción: p y no p.
no_contradiccion <- tabla$p & (!tabla$p)
es_contradiccion(no_contradiccion)
```

### Verificar una equivalencia

Dos fórmulas son lógicamente equivalentes si coinciden en todos los renglones de
la tabla de verdad. Comprobemos con la primera ley de De Morgan, que afirma que
$\neg(p \land q) \equiv \neg p \lor \neg q$.

Observa que calculo cada lado por separado antes de compararlos. Podría escribir
todo en una sola línea, pero con negaciones de por medio el riesgo de precedencia
que ya mencionamos es real, y el código queda además mucho más claro de leer.

```r
lado_izquierdo <- !(tabla$p & tabla$q)
lado_derecho   <- (!tabla$p) | (!tabla$q)

# Comparamos renglón por renglón.
lado_izquierdo == lado_derecho

# Y resumimos el veredicto en un solo valor.
all(lado_izquierdo == lado_derecho)
```

Este patrón —calcular ambos lados, compararlos, resumir con `all()`— es el que
usarás para verificar cualquier equivalencia lógica de aquí en adelante.

## 6. De la lógica al agente

Un agente reactivo simple, en el sentido del Tema I, es aquel que elige su
acción únicamente a partir de la percepción actual, sin memoria de lo ocurrido
antes. Su comportamiento se describe mediante reglas de la forma
*si condición entonces acción*, que no son otra cosa que implicaciones lógicas
con una acción del lado derecho.

Vamos a modelar uno muy sencillo: un robot de limpieza que percibe tres cosas de
su entorno.

```r
# Una percepción es simplemente un conjunto de valores de verdad.
# Usamos una lista porque nos permite nombrar cada sensor.
percepcion_actual <- list(
  hay_obstaculo = FALSE,
  bateria_baja  = TRUE,
  hay_basura    = TRUE
)

percepcion_actual$bateria_baja
```

Las reglas se implementan con una cadena de condicionales. El orden importa:
la primera condición que se cumpla es la que determina la acción, de modo que
estamos codificando también una **prioridad** entre las reglas.

```r
decidir_accion <- function(p) {
  if (p$bateria_baja) {
    "ir a la base de carga"
  } else if (p$hay_obstaculo) {
    "girar"
  } else if (p$hay_basura) {
    "aspirar"
  } else {
    "avanzar"
  }
}

decidir_accion(percepcion_actual)
```

En R, una función devuelve el valor de la última expresión que evalúa, así que no
hace falta escribir `return()`. Aquí el `if` completo es esa última expresión, y
su valor es la cadena correspondiente a la rama que se haya cumplido.

Ahora viene lo interesante. Podemos usar `expand.grid()` otra vez, pero no para
una tabla de verdad sino para generar **todos los estados posibles del entorno**,
y obtener así la tabla completa de comportamiento del agente.

```r
escenarios <- expand.grid(
  hay_obstaculo = c(TRUE, FALSE),
  bateria_baja  = c(TRUE, FALSE),
  hay_basura    = c(TRUE, FALSE)
)

# mapply aplica la función a los tres vectores en paralelo, tomando un
# elemento de cada uno en cada paso. Es la forma idiomática de recorrer
# varias columnas a la vez sin escribir un ciclo.
escenarios$accion <- mapply(
  function(obstaculo, bateria, basura) {
    decidir_accion(list(hay_obstaculo = obstaculo,
                        bateria_baja  = bateria,
                        hay_basura    = basura))
  },
  escenarios$hay_obstaculo,
  escenarios$bateria_baja,
  escenarios$hay_basura
)

escenarios
```

Esta tabla es, en sentido estricto, la especificación completa del agente: ocho
percepciones posibles y la acción que corresponde a cada una. Vale la pena
detenerse a mirarla, porque hace evidentes cosas que las reglas escritas en
prosa esconden. Por ejemplo, cuando la batería está baja el agente ignora por
completo la basura, incluso si la tiene enfrente.

Podemos contar cuántas veces aparece cada acción con `table()`, una función que
usarás mucho a lo largo del semestre.

```r
table(escenarios$accion)
```

# Pregunta 1

El tamaño de la tabla crece de manera exponencial como $2^n$, donde $n$ representa la cantidad de sensores o variables binarias de percepción. En entornos reales con decenas o cientos de estímulos, este crecimiento vuelve intratable la representación explícita del comportamiento del agente, pues requeriría un espacio de memoria y una capacidad computacional desmedidos para generar y almacenar todas las combinaciones posibles. Por esta razón, en lugar de tablas exhaustivas, los agentes prácticos emplean estructuras compactas como árboles de decisión, redes lógicas, bases de conocimiento o algoritmos de aprendizaje que generalizan patrones a partir de la entrada sin evaluar explícitamente todo el espacio de estados.

Pregunta 2

Construir la tabla de verdad completa equivale a realizar una búsqueda no informada a través de un espacio de estados cuyo tamaño es $2^n$ respecto al número de variables proposicionales. Aunque garantiza exhaustividad y corrección, la complejidad temporal en el peor caso es de orden $O(2^n)$, lo que ocasiona una explosión combinatoria al aumentar la cantidad de variables. En lógica proposicional, determinar si una fórmula es satisfacer o tautología corresponde al problema SAT, por lo que resulta mucho más eficiente utilizar algoritmos de inferencia formal, resolución, algoritmos tipo DPLL o heurísticas de búsqueda local que navegan el espacio de estados sin explorar sistemáticamente cada combinación.
