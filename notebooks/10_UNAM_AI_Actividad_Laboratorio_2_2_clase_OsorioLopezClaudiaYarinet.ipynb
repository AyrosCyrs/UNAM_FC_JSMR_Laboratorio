```python
from pyswip import Prolog

# Creamos el objeto Prolog
prolog = Prolog()

# --- DEFINICIÓN DE LA LÓGICA (Simbólica) ---
# Usamos assertz para añadir hechos y reglas
prolog.assertz("padre(juan, maria)")
prolog.assertz("padre(maria, pedro)")
prolog.assertz("abuelo(X, Y) :- padre(X, Z), padre(Z, Y)")

# --- CONSULTA DESDE PYTHON (Razonamiento) ---
# Queremos saber quién es el abuelo de pedro
results = list(prolog.query("abuelo(Abuelo, pedro)"))

if results:
    for res in results:
        print(f"Resultado encontrado: El abuelo de Pedro es {res['Abuelo']}")
else:
    print("No se encontró relación.")
```

**¿Cómo funciona la arquitectura?**

Esta integración permite que Python maneje los datos pesados (o la interfaz) y Prolog se encargue de la inferencia lógica.

- Prolog (El Oráculo): Almacena reglas complejas que serían difíciles de escribir con muchos if/else en Python.

- Python (El Gestor): Recibe el resultado de la consulta y lo procesa (por ejemplo, para mostrarlo en una web o usarlo en una red neuronal).

```python
from pyswip import Prolog

prolog = Prolog()

# --- 1. REPRESENTACIÓN DE LA ONTOLOGÍA ---
# Definimos la jerarquía de clases (Herencia)
prolog.assertz("especie_de(pinguino, ave)")
prolog.assertz("especie_de(ave, animal)")

# --- 2. DEFINICIÓN DE PROPIEDADES Y EXCEPCIONES ---
# REGLA GENERAL: "Las aves vuelan"
# Usamos r"..." (raw string) para que Python no interprete \+ como error
prolog.assertz(r"vuela(X) :- especie_de(X, ave), \+ excepcion_vuela(X)")

# EXCEPCIÓN: "Los pingüinos son una excepción a la regla de volar"
prolog.assertz("excepcion_vuela(pinguino)")

# REGLA ESPECÍFICA: "Los pingüinos no vuelan"
prolog.assertz("no_vuela(X) :- especie_de(X, pinguino)")

# --- 3. INSTANCIAS (INDIVIDUOS) ---
prolog.assertz("es_un(tweety, pinguino)")

# --- 4. MOTOR DE INFERENCIA ---
def razonar_sobre(nombre_individuo):
    print(f"--- Razonando sobre: {nombre_individuo} ---")

    # 1. Hallar su clase
    res_clase = list(prolog.query(f"es_un({nombre_individuo}, Clase)"))
    if not res_clase:
        return "Individuo desconocido."

    clase_directa = res_clase[0]['Clase']

    # 2. Aplicar Principio de Especificidad
    # ¿Es una excepción?
    es_excepcion = bool(list(prolog.query(f"excepcion_vuela({clase_directa})")))

    if es_excepcion:
        # Priorizamos la información de la clase más específica (Pingüino)
        return f"Resultado: {nombre_individuo} es un {clase_directa}. Por ESPECIFICIDAD, sabemos que NO vuela."
    else:
        # Si no es excepción, heredamos la propiedad de la clase superior (Ave)
        vuela = bool(list(prolog.query(f"vuela({clase_directa})")))
        if vuela:
            return f"Resultado: {nombre_individuo} es un {clase_directa}. Por HERENCIA, sabemos que SÍ vuela."

# Ejecución
print(razonar_sobre("tweety"))
```
